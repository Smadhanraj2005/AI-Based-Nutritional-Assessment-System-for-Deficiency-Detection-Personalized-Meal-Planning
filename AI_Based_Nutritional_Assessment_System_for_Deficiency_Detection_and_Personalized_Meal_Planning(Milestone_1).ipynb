```python
from pathlib import Path
from urllib.request import urlretrieve
import pandas as pd
import time
import os

demo = pd.read_sas(DEMO_URL)
dr1tot = pd.read_sas(DR1TOT_URL)
cbc = pd.read_sas(CBC_URL)
vitd = pd.read_sas(VID_URL)
ferritin = pd.read_sas(FERRITIN_URL)

# Display First Five Rows

datasets = {
    "DEMO_L.xpt": demo,
    "DR1TOT_L.xpt": dr1tot,
    "CBC_L.xpt": cbc,
    "VID_L.xpt": vitd,
    "FERTIN_L.xpt": ferritin,
}

for name, df in datasets.items():
    print("=" * 60)
    print(name)
    print("=" * 60)
    print(df.head())
    print()
```

```python
def download_nhanes_files(file_map: dict, dest_dir: Path):

    dest_dir.mkdir(parents=True, exist_ok=True)

    for file_name, url in file_map.items():

        destination = dest_dir / file_name

        if destination.exists() and destination.stat().st_size > 0:
            print(f"[Skip] {file_name}")
            continue

        print(f"[Downloading] {file_name}")

        start = time.time()

        urlretrieve(url, destination)

        print(
            f"Saved ({destination.stat().st_size/1e6:.2f} MB)"
            f" in {time.time()-start:.2f} seconds"
        )
```

```python
# Required Columns
DEMO_CDC_COLS = [
    "SEQN",
    "RIAGENDR",
    "RIDAGEYR",
]

DIET_CDC_COLS = [
    "SEQN",
    "DR1TKCAL",
    "DR1TPROT",
    "DR1TIRON",
    "DR1TCALC",
    "DR1TVD",
    "DR1TVB12",
    "DR1TFOLA",
    "DR1TVARA",
    "DR1TVC",
    "DR1TMAGN",
    "DR1TZINC",
]

CBC_CDC_COLS = [
    "SEQN",
    "LBXHGB",
]

VID_CDC_COLS = [
    "SEQN",
    "LBXVIDMS",
]

FER_CDC_COLS = [
    "SEQN",
    "LBXFER",
]

```

```python
# Select Required Columns


demo = demo[DEMO_CDC_COLS]
diet = diet[DIET_CDC_COLS]
cbc = cbc[CBC_CDC_COLS]
vid = vid[VID_CDC_COLS]
ferritin = ferritin[FER_CDC_COLS]

# Rename Columns


NHANES_RENAME = {
    "SEQN": "person_id",
    "RIAGENDR": "gender",
    "RIDAGEYR": "age_years",
    "DR1TKCAL": "calories_kcal",
    "DR1TPROT": "protein_g",
    "DR1TIRON": "iron_mg",
    "DR1TCALC": "calcium_mg",
    "DR1TVD": "vitamin_d_mcg",
    "DR1TVB12": "vitamin_b12_mcg",
    "DR1TFOLA": "folate_mcg",
    "DR1TVARA": "vitamin_a_mcg",
    "DR1TVC": "vitamin_c_mg",
    "DR1TMAGN": "magnesium_mg",
    "DR1TZINC": "zinc_mg",
    "LBXHGB": "hemoglobin_g_dl",
    "LBXVIDMS": "serum_vitamin_d",
    "LBXFER": "ferritin",
}

demo.rename(columns=NHANES_RENAME, inplace=True)
diet.rename(columns=NHANES_RENAME, inplace=True)
cbc.rename(columns=NHANES_RENAME, inplace=True)
vid.rename(columns=NHANES_RENAME, inplace=True)
ferritin.rename(columns=NHANES_RENAME, inplace=True)

```

```python
import pandas as pd
from pathlib import Path


# Folder containing NHANES files


BASE = Path("/content/drive/MyDrive/Dataset/NHANES")

# Read XPT Files


demo = pd.read_sas(BASE / "DEMO_L.xpt")
diet = pd.read_sas(BASE / "DR1TOT_L.xpt")
cbc = pd.read_sas(BASE / "CBC_L.xpt")
vitd = pd.read_sas(BASE / "VID_L.xpt")
ferritin = pd.read_sas(BASE / "FERTIN_L.xpt")

```

```python
# Columns


DEMO_CDC_COLS = [
    "SEQN",
    "RIAGENDR",
    "RIDAGEYR",
]

DIET_CDC_COLS = [
    "SEQN",
    "DR1TKCAL",
    "DR1TPROT",
    "DR1TIRON",
    "DR1TCALC",
    "DR1TVD",
    "DR1TVB12",
    "DR1TFOLA",
    "DR1TVARA",
    "DR1TVC",
    "DR1TMAGN",
    "DR1TZINC",
]

CBC_CDC_COLS = [
    "SEQN",
    "LBXHGB",
]

VID_CDC_COLS = [
    "SEQN",
    "LBXVIDMS",
]

FER_CDC_COLS = [
    "SEQN",
    "LBXFER",
]

```

```python
# Select columns


demo = demo[DEMO_CDC_COLS]
diet = diet[DIET_CDC_COLS]
cbc = cbc[CBC_CDC_COLS]
vitd = vitd[VID_CDC_COLS]
ferritin = ferritin[FER_CDC_COLS]

# Rename columns


NHANES_RENAME = {
    "SEQN":"person_id",
    "RIAGENDR":"gender",
    "RIDAGEYR":"age_years",
    "DR1TKCAL":"calories_kcal",
    "DR1TPROT":"protein_g",
    "DR1TIRON":"iron_mg",
    "DR1TCALC":"calcium_mg",
    "DR1TVD":"vitamin_d_mcg",
    "DR1TVB12":"vitamin_b12_mcg",
    "DR1TFOLA":"folate_mcg",
    "DR1TVARA":"vitamin_a_mcg",
    "DR1TVC":"vitamin_c_mg",
    "DR1TMAGN":"magnesium_mg",
    "DR1TZINC":"zinc_mg",
    "LBXHGB":"hemoglobin_g_dl",
    "LBXVIDMS":"serum_vitamin_d",
    "LBXFER":"ferritin",
}

demo.rename(columns=NHANES_RENAME,inplace=True)
diet.rename(columns=NHANES_RENAME,inplace=True)
cbc.rename(columns=NHANES_RENAME,inplace=True)
vitd.rename(columns=NHANES_RENAME,inplace=True)
ferritin.rename(columns=NHANES_RENAME,inplace=True)


# Adults only


demo = demo[
    (demo["age_years"]>=18) &
    (demo["age_years"]<=65)
]

```

```python
# Merge
df = demo.merge(diet,on="person_id",how="left")

df = df.merge(cbc,on="person_id",how="left")

df = df.merge(vitd,on="person_id",how="left")

df = df.merge(ferritin,on="person_id",how="left")


# Required columns


COLUMNS_NEEDED = [
    "gender",
    "age_years",
    "calories_kcal",
    "protein_g",
    "iron_mg",
    "calcium_mg",
    "vitamin_d_mcg",
    "vitamin_b12_mcg",
    "folate_mcg",
    "vitamin_a_mcg",
    "vitamin_c_mg",
    "magnesium_mg",
    "zinc_mg",
]

df = df.dropna(subset=COLUMNS_NEEDED).copy()
```

```python
# Labels


def label_iron(row):

    if pd.notna(row["hemoglobin_g_dl"]):

        if row["gender"]==1 and row["hemoglobin_g_dl"]<13:
            return 1

        if row["gender"]==2 and row["hemoglobin_g_dl"]<12:
            return 1

    if pd.notna(row["ferritin"]):
        return int(row["ferritin"]<30)

    return int(row["iron_mg"]<8)


def label_vitd(row):

    if pd.notna(row["serum_vitamin_d"]):
        return int(row["serum_vitamin_d"]<50)

    return int(row["vitamin_d_mcg"]<15)


df["label_iron"] = df.apply(label_iron,axis=1)

df["label_vitamin_d"] = df.apply(label_vitd,axis=1)

df["label_vitamin_b12"] = (df["vitamin_b12_mcg"]<2.4).astype(int)

df["label_calcium"] = (df["calcium_mg"]<1000).astype(int)

df["label_folate"] = (df["folate_mcg"]<400).astype(int)

df["label_vitamin_a"] = (df["vitamin_a_mcg"]<700).astype(int)

df["label_vitamin_c"] = (df["vitamin_c_mg"]<75).astype(int)

df["label_magnesium"] = (df["magnesium_mg"]<320).astype(int)

df["label_zinc"] = (df["zinc_mg"]<8).astype(int)

df["label_protein"] = (df["protein_g"]<50).astype(int)


```

```python
# Output

print("Rows after labels/diet dropna",df.shape)
```

```python
LABEL_COLS = [
    "label_iron",
    "label_vitamin_d",
    "label_vitamin_b12",
    "label_calcium",
    "label_folate",
    "label_vitamin_a",
    "label_vitamin_c",
    "label_magnesium",
    "label_zinc",
    "label_protein",
]

print("Label Balance (0 vs 1)")


for label in LABEL_COLS:

    counts = df[label].value_counts().sort_index()

    negative = counts.get(0, 0)
    positive = counts.get(1, 0)

    total = negative + positive

    positive_pct = (positive / total) * 100

    print(f"\n{label}")
    print(f"(0: {negative}, 1: {positive}) | positive = {positive_pct:.1f}%")
```
