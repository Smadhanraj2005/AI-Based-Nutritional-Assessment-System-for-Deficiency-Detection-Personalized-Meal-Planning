```python
def download_nhanes_files(file_map: dict, dest_dir: Path):

    dest_dir.mkdir(parents=True, exist_ok=True)

    for file_name, url in file_map.items():

        destination = dest_dir / file_name

        if destination.exists() and destination.stat().st_size > 0:
            print(f"[Skip] {file_name}")
            continue

        print(f"[Downloading] {file_name}")

        start = time.time()

        urlretrieve(url, destination)

        print(
            f"Saved ({destination.stat().st_size/1e6:.2f} MB)"
            f" in {time.time()-start:.2f} seconds"
        )
```

```python
import pandas as pd
from pathlib import Path


# Folder containing NHANES files


BASE = Path("/content/drive/MyDrive/Dataset/NHANES")

# Read XPT Files

demo = pd.read_sas(BASE / "DEMO_L.xpt")
diet = pd.read_sas(BASE / "DR1TOT_L.xpt")
cbc = pd.read_sas(BASE / "CBC_L.xpt")
vitd = pd.read_sas(BASE / "VID_L.xpt")
ferritin = pd.read_sas(BASE / "FERTIN_L.xpt")
```

```python
# Columns
DEMO_CDC_COLS = [
    "SEQN",
    "RIAGENDR",
    "RIDAGEYR",
]

DIET_CDC_COLS = [
    "SEQN",
    "DR1TKCAL",
    "DR1TPROT",
    "DR1TIRON",
    "DR1TCALC",
    "DR1TVD",
    "DR1TVB12",
    "DR1TFOLA",
    "DR1TVARA",
    "DR1TVC",
    "DR1TMAGN",
    "DR1TZINC",
]

CBC_CDC_COLS = [
    "SEQN",
    "LBXHGB",
]

VID_CDC_COLS = [
    "SEQN",
    "LBXVIDMS",
]

FER_CDC_COLS = [
    "SEQN",
    "LBXFER",
]

```

```python
# Select columns


demo = demo[DEMO_CDC_COLS]
diet = diet[DIET_CDC_COLS]
cbc = cbc[CBC_CDC_COLS]
vitd = vitd[VID_CDC_COLS]
ferritin = ferritin[FER_CDC_COLS]

# Rename columns


NHANES_RENAME = {
    "SEQN":"person_id",
    "RIAGENDR":"gender",
    "RIDAGEYR":"age_years",
    "DR1TKCAL":"calories_kcal",
    "DR1TPROT":"protein_g",
    "DR1TIRON":"iron_mg",
    "DR1TCALC":"calcium_mg",
    "DR1TVD":"vitamin_d_mcg",
    "DR1TVB12":"vitamin_b12_mcg",
    "DR1TFOLA":"folate_mcg",
    "DR1TVARA":"vitamin_a_mcg",
    "DR1TVC":"vitamin_c_mg",
    "DR1TMAGN":"magnesium_mg",
    "DR1TZINC":"zinc_mg",
    "LBXHGB":"hemoglobin_g_dl",
    "LBXVIDMS":"serum_vitamin_d",
    "LBXFER":"ferritin",
}

demo.rename(columns=NHANES_RENAME,inplace=True)
diet.rename(columns=NHANES_RENAME,inplace=True)
cbc.rename(columns=NHANES_RENAME,inplace=True)
vitd.rename(columns=NHANES_RENAME,inplace=True)
ferritin.rename(columns=NHANES_RENAME,inplace=True)


# Adults only


demo = demo[
    (demo["age_years"]>=18) &
    (demo["age_years"]<=65)
]

```

```python
# Merge
df = demo.merge(diet,on="person_id",how="left")

df = df.merge(cbc,on="person_id",how="left")

df = df.merge(vitd,on="person_id",how="left")

df = df.merge(ferritin,on="person_id",how="left")


# Required columns

COLUMNS_NEEDED = [
    "gender",
    "age_years",
    "calories_kcal",
    "protein_g",
    "iron_mg",
    "calcium_mg",
    "vitamin_d_mcg",
    "vitamin_b12_mcg",
    "folate_mcg",
    "vitamin_a_mcg",
    "vitamin_c_mg",
    "magnesium_mg",
    "zinc_mg",
]

df = df.dropna(subset=COLUMNS_NEEDED).copy()
```

```python
# Labels


def label_iron(row):

    if pd.notna(row["hemoglobin_g_dl"]):

        if row["gender"]==1 and row["hemoglobin_g_dl"]<13:
            return 1

        if row["gender"]==2 and row["hemoglobin_g_dl"]<12:
            return 1

    if pd.notna(row["ferritin"]):
        return int(row["ferritin"]<30)

    return int(row["iron_mg"]<8)


def label_vitd(row):

    if pd.notna(row["serum_vitamin_d"]):
        return int(row["serum_vitamin_d"]<50)

    return int(row["vitamin_d_mcg"]<15)


df["label_iron"] = df.apply(label_iron,axis=1)

df["label_vitamin_d"] = df.apply(label_vitd,axis=1)

df["label_vitamin_b12"] = (df["vitamin_b12_mcg"]<2.4).astype(int)

df["label_calcium"] = (df["calcium_mg"]<1000).astype(int)

df["label_folate"] = (df["folate_mcg"]<400).astype(int)

df["label_vitamin_a"] = (df["vitamin_a_mcg"]<700).astype(int)

df["label_vitamin_c"] = (df["vitamin_c_mg"]<75).astype(int)

df["label_magnesium"] = (df["magnesium_mg"]<320).astype(int)

df["label_zinc"] = (df["zinc_mg"]<8).astype(int)

df["label_protein"] = (df["protein_g"]<50).astype(int)


```

```python
# Output

print("Rows after labels/diet dropna",df.shape)
```

```python
LABEL_COLS = [
    "label_iron",
    "label_vitamin_d",
    "label_vitamin_b12",
    "label_calcium",
    "label_folate",
    "label_vitamin_a",
    "label_vitamin_c",
    "label_magnesium",
    "label_zinc",
    "label_protein",
]

print("Label Balance (0 vs 1)")


for label in LABEL_COLS:

    counts = df[label].value_counts().sort_index()

    negative = counts.get(0, 0)
    positive = counts.get(1, 0)

    total = negative + positive

    positive_pct = (positive / total) * 100

    print(f"\n{label}")
    print(f"(0: {negative}, 1: {positive}) | positive = {positive_pct:.1f}%")
```

```python

# Features (X) and Labels (Y)

from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

# Input Features (Lab values are NOT included to avoid data leakage)
FEATURE_COLS = [
    "gender",
    "age_years",
    "calories_kcal",
    "protein_g",
    "iron_mg",
    "calcium_mg",
    "vitamin_d_mcg",
    "vitamin_b12_mcg",
    "folate_mcg",
    "vitamin_a_mcg",
    "vitamin_c_mg",
    "magnesium_mg",
    "zinc_mg",
]

# Labels
LABEL_COLS = [
    "label_iron",
    "label_vitamin_d",
    "label_vitamin_b12",
    "label_calcium",
    "label_folate",
    "label_vitamin_a",
    "label_vitamin_c",
    "label_magnesium",
    "label_zinc",
    "label_protein",
]

# Create Features (X) and Labels (Y)
X = df[FEATURE_COLS].copy()
Y = df[LABEL_COLS].copy().astype(int)
```

```python
# Train-Test Split (80% Train, 20% Test)
X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size=0.20,
    random_state=42
)

# Preprocessing

preprocess = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
```

```python
# Fit on training data and transform
X_train_scaled = preprocess.fit_transform(X_train)

# Transform test data
X_test_scaled = preprocess.transform(X_test)


# Information


print("X_train:", X_train_scaled.shape, "| Y_train:", Y_train.shape)
print("X_test :", X_test_scaled.shape, "| Y_test :", Y_test.shape)

print("\nFeatures:")
print(FEATURE_COLS)
```

```python
import time
import pandas as pd

from sklearn.ensemble import RandomForestClassifier
from sklearn.multioutput import MultiOutputClassifier
from sklearn.metrics import accuracy_score, f1_score

# Random Forest


rf = MultiOutputClassifier(

    RandomForestClassifier(
        n_estimators=500,
        max_depth=20,
        min_samples_split=5,
        min_samples_leaf=2,
        max_features="sqrt",
        bootstrap=True,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    )

)

print("Training Random Forest...")

start = time.time()

rf.fit(X_train_scaled, Y_train)

print(f"RF done in {time.time()-start:.2f}s")

# Prediction


rf_pred = rf.predict(X_test_scaled)

# Evaluation


rf_accuracy = accuracy_score(
    Y_test.values.ravel(),
    rf_pred.ravel()
)

rf_macro_f1 = f1_score(
    Y_test,
    rf_pred,
    average="macro"
)

rf_micro_f1 = f1_score(
    Y_test,
    rf_pred,
    average="micro"
)

print("\n" + "="*70)
print("SCORES — Random Forest")
print("="*70)

print(f"Accuracy (overall): {rf_accuracy:.4f} ({rf_accuracy*100:.2f}%)")
print(f"F1 macro:           {rf_macro_f1:.4f}")
print(f"F1 micro:           {rf_micro_f1:.4f}")

print("\nPer-label F1")

scores = []

for i, label in enumerate(LABEL_COLS):

    score = f1_score(
        Y_test.iloc[:, i],
        rf_pred[:, i]
    )

    scores.append([label, score])

scores = pd.DataFrame(scores, columns=["nutrient", "f1"])

print(scores)
```

Feature importance

```python
import numpy as np
import pandas as pd

def average_feature_importance(model, feature_names):
    """
    Average feature importance across all estimators
    in a MultiOutputClassifier.
    """

    importances = []

    for est in model.estimators_:
        if hasattr(est, "feature_importances_"):
            importances.append(est.feature_importances_)

    avg = np.mean(importances, axis=0)

    importance_df = pd.DataFrame({
        "Feature": feature_names,
        "Importance": avg
    })

    importance_df = importance_df.sort_values(
        by="Importance",
        ascending=False
    ).reset_index(drop=True)

    return importance_df
```
